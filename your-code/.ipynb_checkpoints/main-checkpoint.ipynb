```python
orders['CustomerID'].nunique() # unique customer count
```

```python
customer_revenue = orders \
    .groupby('CustomerID') \
    .agg({'amount_spent' : 'sum'})
```

```python
customer_revenue.head()
```

```python
customer_revenue['percentile'] = pd.qcut(customer_revenue['amount_spent'], 20, labels=range(5, 101, 5))
```

```python
customer_revenue.head()
```

```python
customer_revenue.value_counts('percentile').sort_index()

# eqial bins as expected
```

```python
customer_revenue['percentile'].dtype
```

```python
customer_revenue.head()
```

```python
def customer_priority(row):
    if 75 <= row['percentile'] <= 95:
        return 'Preffered customer'
    elif row['percentile'] > 95:
        return 'VIP customer'
    else:
        return 'Unclassified'
```

```python
customer_revenue['classification'] = customer_revenue.apply(customer_priority, axis=1)
```

```python
customer_revenue
```

```python
customer_revenue.value_counts('classification')
```

## Q2: How to identify which country has the most VIP Customers?

```python
orders[['CustomerID', 'Country']].drop_duplicates()
```

```python
orders[['CustomerID']].nunique()
```

```python
# a small subset of customers made orders from multiple countries
```

```python
customer_countries = orders.groupby(['CustomerID']).agg({'Country' : 'nunique'})
int_cust = customer_countries.query('Country > 1').index
```

```python
orders \
    .loc[:, ['CustomerID', 'Country']] \
    .drop_duplicates() \
    .query('CustomerID.isin(@int_cust)') \
    .sort_values('CustomerID')
```

```python
customer_country = orders[['CustomerID', 'Country']].drop_duplicates()

customer_country
```

```python
customer_revenue = customer_revenue.reset_index()
```

```python
customer_revenue
```

```python
country_class = pd.merge(
    left=customer_revenue,
    right=customer_country,
    on='CustomerID',
    how='left'
)
```

```python
country_class \
    .query('classification in ["VIP customer"]') \
    .groupby(['Country']).agg({'classification' : 'count'}) \
    .sort_values('classification', ascending=False)
```

## Q3: How to identify which country has the most VIP+Preferred Customers combined?

```python
country_class \
    .query('classification in ["VIP customer", "Preffered customer"]') \
    .groupby(['Country']).agg({'classification' : 'count'}) \
    .sort_values('classification', ascending=False)
```
