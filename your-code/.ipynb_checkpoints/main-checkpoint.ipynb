## Q1: How to identify VIP & Preferred Customers?

We start by importing all the required libraries:

```python
# import required libraries
import numpy as np
import pandas as pd
```

Next, extract and import `Orders` dataset into a dataframe variable called `orders`. Print the head of `orders` to overview the data:

```python
# your code here
orders = pd.read_csv('Orders.csv')
#Se the first data (5) 
orders.head()
```

```python
# As we goonna to separated the CustomerID to clasificate (VIP & Preferent) depending on wath they spend.
#Calculate the total cost per customer

 # "TotalSales = df.groupby('ClientId')['Sales'].apply(lambda x: x.astype(int) / x.astype(int).sum())"

 # "TotalSales = df.groupby('ClientId')['Sales'].sum().reset_index()"

TotalSales = orders.groupby('CustomerID')['amount_spent'].sum ().reset_index()
TotalSales.head()
```

```python
#Calculate the 75th and 95th percentiles.
#From wath amount the custumer (CustumerID) is VIP (> 95th) #separar valores clientes normales,Preferred  y VIP
p75 = TotalSales['amount_spent'].quantile(0.75)
p95 = TotalSales['amount_spent'].quantile(0.95)

def classify(amount) :
    if amount > p95:
        return'VIP'
    elif amount > p75 :
        return 'Prefered'
    else:
        return 'Regular'
TotalSales['Category'] = TotalSales['amount_spent'].apply(classify) 
TotalSales.head()
```
